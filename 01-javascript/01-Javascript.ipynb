## Exercice 2 :
---
En utilisant les méthodes map() et filter(), créer un tableau nomsProduits à partir du tableau produits, contenant uniquement les noms des produits disponibles, c’est-à-dire ceux dont la propriété estDisponible est true.

```javascript
const produits = [
  { nom: "Ordinateur", prix: 7500, estDisponible: true },
  { nom: "Clavier", prix: 250, estDisponible: false },
  { nom: "Souris", prix: 150, estDisponible: true },
  { nom: "Écran", prix: 1800, estDisponible: false }
];
```

````typescript
const produits = [
    { nom: "Ordinateur", prix: 7500, estDisponible: true },
    { nom: "Clavier", prix: 250, estDisponible: true },
    { nom: "Souris", prix: 150, estDisponible: true },
    { nom: "Écran", prix: 1800, estDisponible: false }
  ];

const prodDispo = produits.filter(function(p){
    return p.estDisponible;
})

const nomsProduits = prodDispo.map(p => p.nom);

console.log(nomsProduits)
````

## Exercice 3 :
---
1. Créez un tableau contenant les notes obtenues par un étudiant dans différentes matières : ``` [12, 15, 8, 16, 14, 10, 18, 11, 17, 13, 20, 9] ```.
</br>
2. Utilisez la méthode reduce() pour calculer la somme de toutes les notes.
</br>
3. Utilisez la méthode map() pour créer un nouveau tableau contenant les notes augmentées de 2 points.
</br>
4. Utilisez la syntaxe de destructuring pour extraire la première et la dernière note du tableau original.
</br>
5. Utilisez ces deux notes pour calculer leur produit.
</br>
6. Affichez dans la console la somme des notes, le tableau des notes augmentées et le produit de la première et de la dernière note.

````typescript
const notes = [12, 15, 8, 16, 14, 10, 18, 11, 17, 13, 20, 9];
let somme = notes.reduce((accumulateur, valeurCourante) =>
     accumulateur + valeurCourante
, 0);
console.log(`La somme des nombres est: ${somme}`)
const notesPlus = notes.map(n => n+2)
console.log("Les notes après l'augmentation de 2 sont: ", notesPlus)
const [premiereNote] = notes;
const derniereNote = notes.at(-1);
console.log(`La première note est: ${premiereNote}`);
console.log(`La dernière note est: ${derniereNote}`);
console.log(`Le produit de la première et de la dernière note est: ${premiereNote * derniereNote}`);
````

## Exercice 4 :
---
1. Créez un tableau de produits où chaque produit est représenté par un objet contenant les propriétés **`nom` (string)** et **`prix` (number)**.
```javascript
const produits = [
    { nom: "Ordinateur", prix: 7500 },
    { nom: "Téléphone", prix: 3200 },
    { nom: "Casque", prix: 450 }
];
```

2. Utilisez la méthode **`map`** pour créer un nouveau tableau contenant les noms des produits en **minuscules**.

3. Utilisez la méthode **`reduce`** pour calculer le **prix total** de tous les produits du tableau.

4. Utilisez la **déstructuration** pour échanger les prix de deux produits du tableau (par exemple, le prix du premier et celui du dernier produit).

5. Affichez dans la console :

   * Le tableau des noms des produits en minuscules.
   * Le prix total de tous les produits.
   * Le tableau des produits après l'échange des prix.


````typescript
const produits = [
    { nom: "Ordinateur", prix: 7500 },
    { nom: "Téléphone", prix: 3200 },
    { nom: "Casque", prix: 450 }
];

let nomMin = produits.map(p => p.nom.toLowerCase());
console.log(`${nomMin} \n`);

let total = produits.reduce((acc, p) => acc + p.prix, 0);
console.log(`La somme des prix est: ${total}\n`);

let [first] = produits;
let last = produits.at(-1);

[first.prix, last.prix] = [last.prix, first.prix];
console.log(`Après l'échange, le prix de ${first.nom} est ${first.prix} et le prix de ${last.nom} est ${last.prix}\n`);
console.log(produits);
````
